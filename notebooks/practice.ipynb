```python
#import txt file
df = pd.read_csv('../data/raw/dialogs.txt', sep="\t", header=None, names=['dialog'])
```

```python
import nltk
nltk.download('punkt_tab')

# Now your tokenization will work
df['tokens'] = df['dialog'].apply(nltk.word_tokenize)
```

```python
df.head()
```

```python
#stemming
from nltk.stem import PorterStemmer 
ps = PorterStemmer()
df['stemmed_tokens'] = df['tokens'].apply(lambda tokens: [ps.stem(token) for token in tokens])
df.head()
```

```python
#lamentization
from nltk.stem import WordNetLemmatizer
wnl = WordNetLemmatizer()
df['lemmatized_tokens'] = df['tokens'].apply(lambda tokens: [wnl.lemmatize(token) for token in tokens])
df.head()
```

```python
#postagging
nltk.download('averaged_perceptron_tagger_eng')
df['pos_tags'] = df['tokens'].apply(nltk.pos_tag)
df.head()
```
